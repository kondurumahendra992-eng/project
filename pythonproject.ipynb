visualization using matplotlib:

```python
import matplotlib.pyplot as plt
import os
OUTPUT_DIR = "output"
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

line chart:

```python
plt.figure(figsize=(8, 5))
plt.plot(df.index, df["sepal_length"], marker="o")
plt.title("Sepal Length by Index")
plt.xlabel("Index")
plt.ylabel("Sepal Length")
plt.grid(True)
plt.savefig(os.path.join(OUTPUT_DIR, "line_chart_sepal_length.png"))
plt.show()
species_counts = df["species"].value_counts()
```

bar chart:

```python
plt.figure(figsize=(6, 5))
species_counts.plot(kind="bar")
plt.title("Species Count")
plt.xlabel("Species")
plt.ylabel("Count")
plt.savefig(os.path.join(OUTPUT_DIR, "bar_chart_species_count.png"))
plt.show()
```

histo gram:

```python
plt.figure(figsize=(8, 5))
plt.hist(df["sepal_width"], bins=10)
plt.title("Distribution of sepal Width")
plt.xlabel("sepal Width")
plt.ylabel("Frequency")
plt.savefig(os.path.join(OUTPUT_DIR, "histogram_sepal_width.png"))
plt.show()

print("All charts saved successfully in the 'output' folder.")
```

visualization using ploty:

scatter plot:

```python
import plotly.express as plt
OUTPUT_DIR = "output"
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
import plotly.express as px
from pathlib import Path
```

```python
OUTPUT_DIR = Path("output")
OUTPUT_DIR.mkdir(exist_ok=True)
```

```python
fig1 = px.scatter(
    df,
    x="sepal_length",
    y="petal_length",
    color="species",
    title="sepal length vs petal length"
)
fig1.update_layout(
    xaxis_title="sepal length",
    yaxis_title="petal length"
)
fig1.write_html(f"{OUTPUT_DIR}/scatter_plot.html")
fig1.show()
```

box plot:

```python
fig2 = px.box(
    df,
    x="species",
    y="petal_length",
    title="petal length Distribution by species"
)
fig2.update_layout(
    xaxis_title="species",
    yaxis_title="Petal length"
)
fig2.write_html(f"{OUTPUT_DIR}/box_plot.html")
fig2.show()
```

histogram plot:

```python
fig3 = px.histogram(
    df,
    x="sepal_length",
    title="Distribution of sepal length"
)
fig3.update_layout(
    xaxis_title="sepal length",
    yaxis_title="count"
)
fig3.write_html(f"{OUTPUT_DIR}/histogram.html")
fig3.show()

species_count = df["species"].value_counts().reset_index()
species_count.columns = ["species", "count"]
```

pie chart:

```python
fig4 = px.pie(
    species_count,
    names="species",
    values="count",
    title="species Distribution"
)
fig4.write_html(f"{OUTPUT_DIR}/pie_chart.html")
fig4.show()

print("All Plotly charts saved successfully in the output folder.")
```
